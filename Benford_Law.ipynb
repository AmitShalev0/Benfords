```python
import numpy as np # save me now!
import matplotlib.pyplot as plt
import random
import scipy.special as scs

def benford(data, horizontal=False):
    num = len(data)
    lst = np.sort([str(data[i])[0] for i in range(num)])

    catag = {}
    for data in lst:
        if data in catag:
            catag[data]+=1
        else:
            catag[data]=1
        
    digits = np.array(list(catag.keys()))
    value = np.array(list(catag.values()))/(num/100)

    fig = plt.figure()
    ax = fig.add_subplot(111)

    ax.set_xlabel('First digit')
    ax.set_ylabel('% of occurrence')

    ax.xaxis.label.set_color('b')
    ax.yaxis.label.set_color('b')
    ax.tick_params(axis='x', colors='b')
    ax.tick_params(axis='y', colors='b')
    if horizontal == True:  # Values will display on the bar graph
        for index, val in enumerate(value):
            plt.text(val, index, str(round(val,2)))
        plt.barh(digits,value)
    else:
        plt.bar(digits,value)
    return digits,value
```

```python
# For each function, n is the number of values in the list

def fib_num(n): # Fibonacci numbers
    fiblst=[]
    a=0; c=1
    for i in range (n):
        fiblst.append(c)
        b=c; c+=a; a=b
    return fiblst

def prime_num(n): # Prime numbers
    i=2
    primelst = [2]
    for i in range (2,n):
        for j in range (len(primelst)):
            if i%primelst[j]==0:
                break
        else:
            primelst.append(i)
    primelst.insert(0,1)
    return primelst

def fact_num(n): # Factorial numbers
    return [int(scs.factorial(i)) for i in range(n)]

def expo_num(a,n):
    return [int(a**i) for i in range(n)]

def stirling(n):
    stir = []
    for i in range(1,n):
        e = (i/np.exp(1))**i
        s = np.sqrt(2*np.pi*i)
        a = int(floor(s*e))
        stir.append(a)
    return stir
#    return [int(np.sqrt(2*np.pi*i)(i/np.exp)**i) for i in range(n)]
```
